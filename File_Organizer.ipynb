```python
import os

folders = [
    "Images",
    "Videos",
    "Documents",
    "Excel",
    "Python",
    "Presentations",
    "Others"
]

for folder_name in folders:
    os.makedirs(os.path.join(folder, folder_name), exist_ok=True)

print("Folders created successfully!")
```

```python
import os
import shutil

for file in os.listdir(folder):

    file_path = os.path.join(folder, file)

    # Skip folders
    if os.path.isdir(file_path):
        continue

    extension = os.path.splitext(file)[1].lower()

    if extension in [".jpg", ".jpeg", ".png", ".gif", ".webp"]:
        destination = "Images"

    elif extension in [".mp4", ".mkv", ".avi", ".mov"]:
        destination = "Videos"

    elif extension in [".pdf", ".doc", ".docx", ".txt"]:
        destination = "Documents"

    elif extension in [".xlsx", ".xls", ".csv"]:
        destination = "Excel"

    elif extension in [".ipynb", ".py"]:
        destination = "Python"

    elif extension in [".pptx", ".ppt"]:
        destination = "Presentations"

    else:
        destination = "Others"

    destination_path = os.path.join(folder, destination)

    shutil.move(file_path, os.path.join(destination_path, file))

    print(file, "→", destination)

print("🎉 All files organized!")
```
